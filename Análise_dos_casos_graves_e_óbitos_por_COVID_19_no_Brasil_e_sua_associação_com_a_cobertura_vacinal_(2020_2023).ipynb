## 1. Contextualização do problema

A pandemia de COVID-19 produziu impactos significativos sobre os sistemas de saúde e sobre a população brasileira. Ao longo do período pandêmico, diferentes regiões do país apresentaram comportamentos distintos em relação à ocorrência de casos graves, óbitos e volume de doses de vacina aplicadas.

No Brasil, a vigilância de casos graves de doenças respiratórias é realizada pelo Sistema de Informação da Vigilância Epidemiológica da Gripe (SIVEP-Gripe), que registra os casos classificados como Síndrome Respiratória Aguda Grave (SRAG), incluindo aqueles causados especificamente pela COVID-19. De forma complementar, o Ministério da Saúde disponibiliza, por meio do painel Vacinômetro COVID-19, dados sobre a aplicação de doses de vacina no país.

Este projeto propõe utilizar técnicas de análise de dados sobre essas duas fontes para investigar a evolução dos casos graves e óbitos por COVID-19, bem como o comportamento da vacinação no Brasil, considerando diferenças regionais, temporais, demográficas e de grupos prioritários.

É importante destacar duas delimitações de escopo, identificadas durante o entendimento inicial dos dados:

1. Por se tratar de uma base de vigilância de casos graves e hospitalizados, os indicadores de "casos" obtidos neste estudo referem-se aos casos de SRAG por COVID-19, não à totalidade dos casos confirmados da doença (incluindo os leves e moderados, não capturados por esse sistema).
2. As bases de casos/óbitos e de vacinação utilizadas neste projeto são agregadas (por semana epidemiológica, unidade federativa, município e faixa etária, no caso dos casos/óbitos; e por unidade federativa, data, grupo prioritário, faixa etária e gênero, no caso da vacinação), e não trazem o registro individual de cada paciente. Por esse motivo, não é possível identificar diretamente se um paciente específico que teve SRAG por COVID-19 havia sido vacinado. As análises que relacionam vacinação e letalidade neste projeto são, portanto, análises agregadas, ou seja, comparam o comportamento de grupos (regiões, faixas etárias, períodos), não de indivíduos, e não permitem conclusões causais diretas.

## 2. Problema de negócio

Durante a pandemia de COVID-19, gestores públicos de saúde enfrentaram o desafio de acompanhar a evolução da letalidade por casos graves da doença e de planejar a distribuição da vacinação em diferentes regiões e grupos populacionais. Entretanto, diferenças na velocidade de vacinação, no perfil demográfico dos pacientes e na evolução temporal da pandemia podem produzir comportamentos distintos entre as regiões brasileiras.

Dessa forma, o problema de negócio investigado neste projeto consiste em compreender como os casos graves e os óbitos por COVID-19 evoluíram ao longo do período analisado, como se comportou a aplicação de doses de vacina no país e se há indícios de associação entre esses dois fenômenos, considerando diferenças regionais, temporais e demográficas.

O "negócio", neste contexto, corresponde a gestores públicos de saúde (Ministério da Saúde, secretarias estaduais/municipais), que precisam compreender esses padrões para apoiar decisões relacionadas à priorização de grupos vulneráveis e ao planejamento de políticas públicas de vacinação e resposta a emergências sanitárias.

## 2. Carregamento e inventário inicial das bases

Antes de realizar qualquer transformação, é necessário conhecer a estrutura dos arquivos de entrada. O inventário abaixo permite verificar quais arquivos estão disponíveis, suas dimensões e suas colunas. Essa etapa também cria uma referência objetiva para comparar a situação dos dados antes e depois do tratamento.


```python
# Inventário dos arquivos de casos/óbitos
arquivos_casos = sorted(
    arquivo for arquivo in os.listdir(pasta_obitos)
    if arquivo.startswith("covid_") and arquivo.endswith(".xlsx")
)

if not arquivos_casos:
    raise FileNotFoundError(
        f"Nenhum arquivo 'covid_*.xlsx' foi encontrado em: {pasta_obitos}"
    )

inventario_casos = []
for arquivo in arquivos_casos:
    caminho = os.path.join(pasta_obitos, arquivo)
    temp = pd.read_excel(caminho, sheet_name="Completo")
    inventario_casos.append({
        "Arquivo": arquivo,
        "Linhas": len(temp),
        "Colunas": len(temp.columns),
        "Colunas disponíveis": ", ".join(map(str, temp.columns))
    })

df_inventario_casos = pd.DataFrame(inventario_casos)
display(df_inventario_casos)
```

```python
# Inventário das bases de vacinação de entrada
arquivos_vacinacao = [
    "por_estado_F.xlsx",
    "por_estado_M.xlsx",
    "por_data_vacina.xlsx",
    "por_grupo.xlsx",
    "por_faixa_etaria.xlsx"
]

inventario_vacinacao = []
for arquivo in arquivos_vacinacao:
    caminho = os.path.join(pasta_vacinacao, arquivo)
    if os.path.exists(caminho):
        temp = pd.read_excel(caminho)
        inventario_vacinacao.append({
            "Arquivo": arquivo,
            "Linhas": len(temp),
            "Colunas": len(temp.columns),
            "Colunas disponíveis": ", ".join(map(str, temp.columns))
        })
    else:
        inventario_vacinacao.append({
            "Arquivo": arquivo,
            "Linhas": None,
            "Colunas": None,
            "Colunas disponíveis": "ARQUIVO NÃO ENCONTRADO"
        })

df_inventario_vacinacao = pd.DataFrame(inventario_vacinacao)
display(df_inventario_vacinacao)
```

## 3. Consolidação das bases de dados

## 3.1. Consolidação dos Casos e Óbitos


### Decisão metodológica

Como os arquivos anuais de casos e óbitos apresentam a mesma finalidade e uma estrutura semelhante, eles são reunidos em uma única tabela. Para isso, o código percorre os arquivos disponíveis, lê a planilha “Completo” de cada arquivo e acrescenta o nome do arquivo em uma nova coluna chamada Arquivo.

Essa informação é mantida inicialmente porque será utilizada posteriormente para identificar o ano de referência de cada registro. Depois, todos os arquivos são concatenados em uma única base, mantendo os registros de forma contínua.



```python
# CONSOLIDAÇÃO DAS BASES DE CASOS E ÓBITOS

dados = []
for arquivo in arquivos_casos:
    caminho = os.path.join(pasta_obitos, arquivo)
    df_temp = pd.read_excel(caminho, sheet_name="Completo")
    df_temp["Arquivo"] = arquivo
    dados.append(df_temp)

consolidado = pd.concat(dados, ignore_index=True)

arquivo_saida_consolidado = os.path.join(pasta_obitos, "Consolidado.xlsx")
consolidado.to_excel(arquivo_saida_consolidado, index=False)

print(f"Arquivos consolidados: {len(arquivos_casos):,}")
print(f"Registros: {len(consolidado):,}")
print(f"Colunas: {len(consolidado.columns):,}")
display(consolidado.head())
```

```python
# Validação da consolidação inicial
print("Dimensão:", consolidado.shape)
print("Duplicidades exatas:", consolidado.duplicated().sum())
print("Colunas:", list(consolidado.columns))
print("\nTipos iniciais:")
display(consolidado.dtypes.to_frame("Tipo"))
```

## 3.1.3. Padronização da base de Casos e Óbitos

Nesta etapa, a base consolidada é ajustada para deixar as informações em um formato único e adequado às próximas etapas do projeto.

A padronização seguirá a seguinte lógica de desenvolvimento:

1. extrair o ano do arquivo de origem;
2. completar o município a partir do código quando a estrutura dos dados permitir;
3. tratar registros com `BR` na UF utilizando informações válidas associadas ao código;
4. converter a semana epidemiológica para mês;
5. padronizar nomes das variáveis;
6. padronizar faixa etárias;
7. remover coluna `Regional de Saúde`uma vez que é informação não relevante para este projeto;
8. remover os registros que permanecerem sem a informação de município após todas as etapas de preenchimento;
9. definir o código do município como índice.



```python
# PADRONIZAÇÃO DA BASE DE CASOS E ÓBITOS

df = consolidado.copy()

# Função para padronizar e agrupar as faixas etárias nas 7 categorias requeridas
# Função corrigida para padronizar e agrupar as faixas etárias
def ajustar_faixa_etaria(faixa):
    if pd.isna(faixa):
        return "Sem informação"

    faixa_str = str(faixa).strip().lower()

    if faixa_str in ["-", "", "nan", "none", "sem informação", "sem informacao"]:
        return "Sem informação"

    # 1. < 02 anos
    if faixa_str in ["< 1 ano", "<1 ano", "1 ano", "< 2 anos", "<2 anos", "< 02 anos", "<02 anos", "06 meses a 02 anos", "6 meses a 2 anos", "6 meses a 02 anos"]:
        return "< 02 anos"

    # 2. 02 a 04 anos (Usando correspondência exata para evitar capturar '14 anos' ou '64 anos')
    if faixa_str in ["02 a 04 anos", "2 a 4 anos", "03 a 04 anos", "3 a 4 anos", "2 anos", "3 anos", "4 anos"]:
        return "02 a 04 anos"

    # 3. 05 a 14 anos
    if faixa_str in ["05 a 14 anos", "5 a 14 anos", "05 a 11 anos", "5 a 11 anos", "12 a 17 anos", "5 a 9 anos", "10 a 14 anos"]:
        return "05 a 14 anos"

    # 4. 15 a 49 anos
    if faixa_str in [
        "15 a 49 anos", "18 a 19 anos", "20 a 24 anos", "25 a 29 anos",
        "30 a 34 anos", "35 a 39 anos", "40 a 44 anos", "45 a 49 anos",
        "15 a 19 anos", "20 a 29 anos", "30 a 39 anos", "40 a 49 anos"
    ]:
        return "15 a 49 anos"

    # 5. 50 a 64 anos
    if faixa_str in ["50 a 64 anos", "50 a 54 anos", "55 a 59 anos", "60 a 64 anos", "50 a 59 anos"]:
        return "50 a 64 anos"

    # 6. 65 anos ou +
    if faixa_str in ["65 anos ou +", "65 a 69 anos", "70 a 74 anos", "75 a 79 anos", "80 anos ou mais", "80 anos ou +", "65 anos ou mais", "70 a 79 anos"]:
        return "65 anos ou +"

    return str(faixa).strip()

# 1 — Extrair ano do nome do arquivo
anos = (
    df["Arquivo"].astype("string").str.extract(r"(\d+)", expand=False)
)
anos = anos.mask(anos.str.len().eq(2), "20" + anos)
df["Arquivo"] = pd.to_numeric(anos, errors="coerce").astype("Int64")

# 2 — Completar Município pelo Código, preservando a regra do tratamento original
if "Codigo" in df.columns and "Município" in df.columns:
    df["Município"] = (
        df.groupby("Codigo", sort=False)["Município"]
        .transform("first")
    )

# 3 — Tratar registros BR na UF
coluna_uf_original = "UF *** Contém BR - CUIDADO ***"
if coluna_uf_original in df.columns:
    df[coluna_uf_original] = df[coluna_uf_original].astype("string").str.strip()
    uf_valida_por_codigo = (
        df.loc[
            ~df[coluna_uf_original].isin(["BR", "nan", "None"]),
            ["Codigo", coluna_uf_original]
        ]
        .drop_duplicates("Codigo")
        .set_index("Codigo")[coluna_uf_original]
    )
    mask_br = df[coluna_uf_original].eq("BR")
    qtd_br_antes = int(mask_br.sum())
    df.loc[mask_br, coluna_uf_original] = df.loc[mask_br, "Codigo"].map(uf_valida_por_codigo)
else:
    qtd_br_antes = 0

# 4 — Converter semana epidemiológica para mês
mapeamento_se_mes = {
    1:"Janeiro",2:"Janeiro",3:"Janeiro",4:"Janeiro",5:"Janeiro",
    6:"Fevereiro",7:"Fevereiro",8:"Fevereiro",9:"Fevereiro",
    10:"Março",11:"Março",12:"Março",13:"Março",
    14:"Abril",15:"Abril",16:"Abril",17:"Abril",18:"Abril",
    19:"Maio",20:"Maio",21:"Maio",22:"Maio",
    23:"Junho",24:"Junho",25:"Junho",26:"Junho",
    27:"Julho",28:"Julho",29:"Julho",30:"Julho",31:"Julho",
    32:"Agosto",33:"Agosto",34:"Agosto",35:"Agosto",
    36:"Setembro",37:"Setembro",38:"Setembro",39:"Setembro",
    40:"Outubro",41:"Outubro",42:"Outubro",43:"Outubro",44:"Outubro",
    45:"Novembro",46:"Novembro",47:"Novembro",48:"Novembro",
    49:"Dezembro",50:"Dezembro",51:"Dezembro",52:"Dezembro",53:"Dezembro"
}

df["Semana Epidemiológica"] = pd.to_numeric(
    df["Semana Epidemiológica"], errors="coerce"
).map(mapeamento_se_mes)

# 5 — Padronizar nomes
rename_map = {
    "Semana Epidemiológica":"Mês",
    "Arquivo":"Ano",
    coluna_uf_original:"UF",
    "casos":"Casos",
    "obitos":"Óbitos"
}
df.rename(columns=rename_map, inplace=True)
df.drop(columns=["Regional de Saúde"], errors="ignore", inplace=True)

# 5.1 — Aplicar padronização da Faixa Etária
if "Faixa Etária" in df.columns:
    df["Faixa Etária"] = df["Faixa Etária"].apply(ajustar_faixa_etaria)

# ETAPA: Remover registros com Município vazio/nulo ao final
if "Município" in df.columns:
    df["Município"] = df["Município"].astype("string").str.strip()
    df = df[~df["Município"].isin(["", "nan", "None"]) & df["Município"].notna()]

# 6 — Organizar estrutura
colunas_base = ["Codigo", "UF", "Município", "Faixa Etária", "Ano", "Mês", "Casos", "Óbitos"]
colunas_existentes = [c for c in colunas_base if c in df.columns]
df_ajustado_1 = df[colunas_existentes].copy().set_index("Codigo")

arquivo_saida_ajuste1 = os.path.join(pasta_obitos, "Consolidado_Ajuste_1.xlsx")
df_ajustado_1.to_excel(arquivo_saida_ajuste1, index=True)

print(f"Registros após padronização: {len(df_ajustado_1):,}")
print(f"Registros BR identificados antes da correção: {qtd_br_antes:,}")
display(df_ajustado_1.head(10))
```

## 3.1.4. Consolidação da granularidade de Casos e Óbitos

Nesta etapa, é definida a unidade de observação que será utilizada na base final. Cada registro deve representar uma combinação única de `Código + UF + Município + Ano + Mês + Faixa Etária`.

Quando existem vários registros para a mesma combinação, eles não são simplesmente descartados. Eles são agregados por soma de `Casos` e `Óbitos`. Essa decisão transforma registros que pertencem à mesma unidade analítica em uma única observação.

Depois da agregação, a Região é criada a partir da UF. Para isso, é utilizado um mapeamento que relaciona cada uma das 27 unidades federativas às cinco regiões brasileiras: Norte, Nordeste, Centro-Oeste, Sudeste e Sul.


```python
# AGRUPAMENTO E CONSOLIDAÇÃO

df_agrupamento = df_ajustado_1.reset_index()

registros_antes_agregacao = len(df_agrupamento)
chaves_duplicadas = df_agrupamento.duplicated(
    subset=["Codigo", "UF", "Município", "Ano", "Mês", "Faixa Etária"],
    keep=False
).sum()

mapeamento_regioes = {
    "AM":"Norte","RR":"Norte","AP":"Norte","PA":"Norte","TO":"Norte","RO":"Norte","AC":"Norte",
    "MA":"Nordeste","PI":"Nordeste","CE":"Nordeste","RN":"Nordeste","PE":"Nordeste","PB":"Nordeste","SE":"Nordeste","AL":"Nordeste","BA":"Nordeste",
    "MT":"Centro-Oeste","MS":"Centro-Oeste","GO":"Centro-Oeste","DF":"Centro-Oeste",
    "SP":"Sudeste","RJ":"Sudeste","MG":"Sudeste","ES":"Sudeste",
    "PR":"Sul","RS":"Sul","SC":"Sul"
}

df_ajustado_2 = (
    df_agrupamento
    .groupby(
        ["Codigo", "UF", "Município", "Ano", "Mês", "Faixa Etária"],
        as_index=False, sort=False, dropna=False
    )[["Casos", "Óbitos"]]
    .sum()
    .set_index("Codigo")
)

df_ajustado_2["Região"] = df_ajustado_2["UF"].map(mapeamento_regioes)
df_ajustado_2 = df_ajustado_2[["UF","Região","Município","Faixa Etária","Ano","Mês","Casos","Óbitos"]]
df_ajustado_2 = df_ajustado_2.sort_values(by=["UF","Município"], ascending=[True,True])


# Verificação e tratamento de registros sem Região resolvível
# (ocorre quando o UF permanece como 'BR' após a etapa 3.1.3, por não existir
# nenhum outro registro do mesmo Código com uma UF válida para preencher o valor)
qtd_regiao_ausente = df_ajustado_2["Região"].isna().sum()
if qtd_regiao_ausente > 0:
    ufs_nao_mapeadas = df_ajustado_2.loc[df_ajustado_2["Região"].isna(), "UF"].unique()
    print(f"Registros sem Região resolvível: {qtd_regiao_ausente:,}")
    print(f"Valores de UF não mapeados: {ufs_nao_mapeadas}")
    # Esses registros representam a linha de total nacional ('BR') que não pôde ser
    # associada a nenhuma Unidade Federativa. Como não é possível atribuí-los a uma
    # região real, e mantê-los distorceria qualquer agregação regional, são removidos.
    df_ajustado_2 = df_ajustado_2[df_ajustado_2["Região"].notna()].copy()
    print(f"Registros removidos por ausência de Região: {qtd_regiao_ausente:,}")

arquivo_saida_ajuste2 = os.path.join(pasta_obitos, "Consolidado_Ajuste_2.xlsx")
df_ajustado_2.to_excel(arquivo_saida_ajuste2, index=True)

print(f"Registros antes da agregação: {registros_antes_agregacao:,}")
print(f"Registros envolvidos em chaves repetidas: {chaves_duplicadas:,}")
print(f"Registros após agregação: {len(df_ajustado_2):,}")
display(df_ajustado_2.head(10))
```

```python
# Validação da granularidade final de casos/óbitos
chaves_finais = df_ajustado_2.reset_index()[
    ["Codigo","UF","Município","Ano","Mês","Faixa Etária"]
]

print("Duplicidades na granularidade final:", chaves_finais.duplicated().sum())
print("Casos negativos:", (pd.to_numeric(df_ajustado_2["Casos"], errors="coerce") < 0).sum())
print("Óbitos negativos:", (pd.to_numeric(df_ajustado_2["Óbitos"], errors="coerce") < 0).sum())
print("Óbitos maiores que casos:", (
    pd.to_numeric(df_ajustado_2["Óbitos"], errors="coerce") >
    pd.to_numeric(df_ajustado_2["Casos"], errors="coerce")
).sum())
```

## 3.3. Preparação da vacinação por data

A visão temporal da vacinação é necessária para a pergunta P3. A competência é convertida em data e, posteriormente, em ano e mês. Os registros são agregados por período.

A soma é realizada antes de qualquer formatação visual, mantendo `Total de Doses Aplicadas` como variável numérica.


```python
# VACINAÇÃO POR DATA
caminho_vacina_data = os.path.join(pasta_vacinacao, "por_data_vacina.xlsx")
arquivo_saida_vacina_data = os.path.join(pasta_vacinacao, "vacina_data_consolidada.xlsx")

df_vacina_data = pd.read_excel(caminho_vacina_data)

df_vacina_data["Competencia Vacina"] = pd.to_datetime(df_vacina_data["Competencia Vacina"], errors="coerce")

df_vacina_data["Ano"] = df_vacina_data["Competencia Vacina"].dt.year
df_vacina_data["Num_Mes"] = df_vacina_data["Competencia Vacina"].dt.month

# Filtra apenas os anos de 2021 a 2023
df_vacina_data = df_vacina_data[df_vacina_data["Ano"].between(2021, 2023)]

mapeamento_meses = {
    1: "Janeiro", 2: "Fevereiro", 3: "Março", 4: "Abril",
    5: "Maio", 6: "Junho", 7: "Julho", 8: "Agosto",
    9: "Setembro", 10: "Outubro", 11: "Novembro", 12: "Dezembro"
}

df_vacina_data["Mês"] = df_vacina_data["Num_Mes"].map(mapeamento_meses)

# Identifica a primeira coluna numérica de doses, mantendo a regra original.
colunas_excluir = {"Ano", "Num_Mes"}
colunas_numericas = [
    coluna
    for coluna in df_vacina_data.select_dtypes(include="number").columns
    if coluna not in colunas_excluir
]

if not colunas_numericas:
    raise ValueError("Nenhuma coluna numérica de vacinas foi encontrada.")

coluna_vacinas = colunas_numericas[0]

print(
    f"Somando vacinas pela coluna '{coluna_vacinas}' "
    "agrupadas por Ano e Mês..."
)

df_vacina_data_consolidado = (
    df_vacina_data
    .groupby(
        ["Ano", "Num_Mes", "Mês"],
        as_index=False,
        sort=False,
        dropna=False
    )[coluna_vacinas]
    .sum()
    .sort_values(["Ano", "Num_Mes"], kind="stable")
    .drop(columns="Num_Mes")
    .rename(columns={coluna_vacinas: "Total de Doses Aplicadas"})
)

# CORREÇÃO: Limpeza, tratamento nulo e tipo numérico mantidos sem converter para texto.
df_vacina_data_consolidado["Total de Doses Aplicadas"] = (
    pd.to_numeric(
        df_vacina_data_consolidado["Total de Doses Aplicadas"],
        errors="coerce"
    )
    .fillna(0)
    .round()
    .astype("int64")
)

df_vacina_data_consolidado.set_index("Ano", inplace=True)

df_vacina_data_consolidado.to_excel(
    arquivo_saida_vacina_data,
    index=True
)

display(df_vacina_data_consolidado.head(12))
```

```python
# Validação da vacinação por data

print(
    "Anos da vacinação:",
    sorted(
        df_vacina_data["Ano"]
        .dropna()
        .unique()
        .tolist()
    )
)

print(
    "Competências inválidas na origem:",
    df_vacina_data["Competencia Vacina"].isna().sum()
)

print(
    "Doses negativas:",
    (
        pd.to_numeric(
            df_vacina_data["Total de Doses Aplicadas"],
            errors="coerce"
        )
        < 0
    ).sum()
)

print(
    "Duplicidades Ano + Mês:",
    df_vacina_data.duplicated(
        subset=["Ano", "Mês"]
    ).sum()
)
```

## 3.4. Preparação da vacinação por grupo prioritário

A pergunta P7 requer a identificação dos grupos que concentram maior volume de doses. A coluna de categoria é padronizada para que valores representados por `-` não permaneçam como uma categoria ambígua. Essa substituição somente deve ser interpretada como “Sem informação” quando o significado estiver de acordo com a estrutura da fonte.

Depois da padronização, as doses são agregadas por categoria e mantidas como números.


```python
# VACINAÇÃO POR GRUPO/CATEGORIA
caminho_vacina_grupo = os.path.join(pasta_vacinacao, "por_grupo.xlsx")
df_vacina_grupo_origem = pd.read_excel(caminho_vacina_grupo)

df_vacina_grupo_origem["Categoria_Nome"] = (
    df_vacina_grupo_origem["Categoria_Nome"].replace("-", "Sem informação")
)

colunas_numericas = [
    c for c in df_vacina_grupo_origem.select_dtypes(include="number").columns
    if c != "Doses"
]

if not colunas_numericas:
    raise ValueError("Nenhuma coluna numérica de vacinação foi encontrada.")

coluna_vacinas = colunas_numericas[0]

df_vacina_grupo = (
    df_vacina_grupo_origem
    .groupby(
        "Categoria_Nome",
        as_index=False,
        sort=False,
        dropna=False
    )[coluna_vacinas]
    .sum()
    .sort_values(
        coluna_vacinas,
        ascending=False,
        kind="stable"
    )
    .rename(
        columns={
            "Categoria_Nome": "Grupo Prioridade",
            coluna_vacinas: "Total de Doses Aplicadas"
        }
    )
)

df_vacina_grupo["Total de Doses Aplicadas"] = pd.to_numeric(
    df_vacina_grupo["Total de Doses Aplicadas"],
    errors="coerce"
).fillna(0).round().astype("int64")

# Define 'Grupo Prioridade' como INDEX
df_vacina_grupo.set_index(
    "Grupo Prioridade",
    inplace=True
)

arquivo_saida_vacina_grupo = os.path.join(
    pasta_vacinacao,
    "vacina_grupo_consolidada.xlsx"
)

# Exportação (index=True para salvar 'Grupo Prioridade' como índice)
df_vacina_grupo.to_excel(
    arquivo_saida_vacina_grupo,
    index=True
)

display(df_vacina_grupo)
```

## 3.5. Preparação da vacinação por faixa etária

A base de vacinação possui faixas etárias mais detalhadas do que a base de casos/óbitos. Para permitir a análise prevista na P6, as faixas da vacinação são reagrupadas nas categorias utilizadas pela base de casos/óbitos.

A regra de reclassificação deve ser explícita e aplicada antes da agregação. O procedimento abaixo preserva a lógica do notebook de tratamento, mas mantém as doses como valores numéricos na base analítica final.


```python
# VACINAÇÃO POR FAIXA ETÁRIA


# 1. Carregamento dos dados
caminho_vacina_idade = os.path.join(pasta_vacinacao, "por_faixa_etaria.xlsx")
df_vacina_idade_origem = pd.read_excel(caminho_vacina_idade)

# 2. Mapeamento direto e sem ambiguidades com base no DE/PARA da imagem
mapa_faixas = {
    "6 meses a 2 anos": "< 02 anos",
    "3 a 4 anos": "02 a 04 anos",
    "5 a 11 anos": "05 a 14 anos",
    "12 a 17 anos": "15 a 49 anos",
    "18 a 19 anos": "15 a 49 anos",
    "20 a 24 anos": "15 a 49 anos",
    "25 a 29 anos": "15 a 49 anos",
    "30 a 34 anos": "15 a 49 anos",
    "35 a 39 anos": "15 a 49 anos",
    "40 a 44 anos": "15 a 49 anos",
    "45 a 49 anos": "15 a 49 anos",
    "50 a 54 anos": "50 a 64 anos",
    "55 a 59 anos": "50 a 64 anos",
    "60 a 64 anos": "50 a 64 anos",
    "65 a 69 anos": "65 anos ou +",
    "70 a 74 anos": "65 anos ou +",
    "75 a 79 anos": "65 anos ou +",
    "80 anos ou mais": "65 anos ou +"
}

# Aplicação da consolidação
df_vacina_idade_origem["Faixa Etária Consolidada"] = (
    df_vacina_idade_origem["Faixa Etária"].map(mapa_faixas).fillna("Sem informação")
)

# Identifica a coluna numérica de doses agregadas
coluna_vacinas = "Total de Doses Aplicadas"

# 3. Agrupamento e Soma
df_vacina_idade = (
    df_vacina_idade_origem
    .groupby("Faixa Etária Consolidada", as_index=False, dropna=False)[coluna_vacinas]
    .sum()
    .rename(columns={"Faixa Etária Consolidada": "Faixa Etária"})
)

# 4. Ordenação estrita conforme granulidade oficial
ordem_faixas = [
    "< 02 anos",
    "02 a 04 anos",
    "05 a 14 anos",
    "15 a 49 anos",
    "50 a 64 anos",
    "65 anos ou +",
    "Sem informação"
]

mapa_ordem = {faixa: i for i, faixa in enumerate(ordem_faixas)}
df_vacina_idade["ordem"] = df_vacina_idade["Faixa Etária"].map(mapa_ordem).fillna(999)
df_vacina_idade = df_vacina_idade.sort_values("ordem").drop(columns="ordem").reset_index(drop=True)

# Garantia de tipo numérico int64
df_vacina_idade["Total de Doses Aplicadas"] = pd.to_numeric(
    df_vacina_idade["Total de Doses Aplicadas"], errors="coerce"
).fillna(0).round().astype("int64")

# 5. Configura 'Faixa Etária' como índice e salva no Excel
df_vacina_idade.set_index("Faixa Etária", inplace=True)

arquivo_saida_vacina_idade = os.path.join(pasta_vacinacao, "vacina_idade_consolidada.xlsx")
df_vacina_idade.to_excel(arquivo_saida_vacina_idade, index=True)

display(df_vacina_idade)
```

```python
# Validação da consolidação por faixa etária

# 1. Lista de faixas etárias presentes
print("Faixas finais:")
print(df_vacina_idade.index.tolist())

# 2. Verificação de duplicidades no índice
print("Duplicidades:", df_vacina_idade.index.duplicated().sum())

# 3. Verificação de doses negativas
print("Doses negativas:", (df_vacina_idade["Total de Doses Aplicadas"] < 0).sum())

# 4. Tipos das colunas restantes na tabela
print("Tipos finais:")
display(df_vacina_idade.dtypes.to_frame("Tipo"))
```

## 2. Estatísticas descritivas das bases analíticas

Para casos e óbitos, são observadas medidas como média, mediana, desvio-padrão, mínimo, máximo e quartis. Para vacinação, a mesma lógica é aplicada ao volume mensal de doses.

A utilização dessas medidas é importante porque valores absolutos isolados não mostram como os dados se distribuem. A comparação entre média e mediana, por exemplo, pode indicar concentração ou assimetria, enquanto a amplitude e o desvio-padrão ajudam a avaliar a variabilidade dos registros.

## P1. Como evoluíram os casos graves (SRAG) e os óbitos por COVID-19 no Brasil entre 2020 e 2023?

A P1 analisa a evolução temporal da SRAG por COVID-19 entre 2020 e 2023 por meio de dados mensais e anuais, permitindo visualizar a distribuição conjunta de casos graves e óbitos e identificar picos de maior concentração sem antecipar justificativas causais.

> **Observação metodológica:** Como 2020 é o ano inicial da série analisada e não possui um período anterior para comparação, sua variação percentual é indicada como "Não se aplica", evitando o uso incorreto do valor 0%.

```python
# Agregação mensal de casos e óbitos
p1_mensal = (
    df_casos_obitos.groupby(["Ano", "Mês"], as_index=False)[["Casos", "Óbitos"]]
    .sum()
)
p1_mensal["Mês_num"] = p1_mensal["Mês"].map(meses_ordem)
p1_mensal["Data"] = pd.to_datetime(
    p1_mensal["Ano"].astype(str) + "-" + p1_mensal["Mês_num"].astype(str) + "-01",
    errors="coerce"
)
p1_mensal = p1_mensal.sort_values("Data")

p1_anual = (
    p1_mensal.groupby("Ano", as_index=False)[["Casos", "Óbitos"]].sum()
)
p1_anual["Participação dos casos (%)"] = p1_anual["Casos"] / p1_anual["Casos"].sum() * 100
p1_anual["Participação dos óbitos (%)"] = p1_anual["Óbitos"] / p1_anual["Óbitos"].sum() * 100
# Variação percentual em relação ao ano anterior
p1_anual["Variação casos (%)"] = p1_anual["Casos"].pct_change() * 100
p1_anual["Variação óbitos (%)"] = p1_anual["Óbitos"].pct_change() * 100

# Colunas para apresentação, deixando explícito que 2020 não possui
# período anterior dentro do recorte analisado
p1_anual["Variação casos (apresentação)"] = (
    p1_anual["Variação casos (%)"]
    .round(2)
    .astype("string")
)

p1_anual["Variação óbitos (apresentação)"] = (
    p1_anual["Variação óbitos (%)"]
    .round(2)
    .astype("string")
)

p1_anual.loc[
    p1_anual["Ano"].eq(p1_anual["Ano"].min()),
    ["Variação casos (apresentação)", "Variação óbitos (apresentação)"]
] = "Não se aplica"

print("\n")
print("Resumo anual:")

display(
    p1_anual[
        [
            "Ano",
            "Casos",
            "Óbitos",
            "Participação dos casos (%)",
            "Participação dos óbitos (%)",
            "Variação casos (apresentação)",
            "Variação óbitos (apresentação)"
        ]
    ]
)

# Picos mensais
pico_casos = p1_mensal.loc[p1_mensal["Casos"].idxmax()]
pico_obitos = p1_mensal.loc[p1_mensal["Óbitos"].idxmax()]
print("\n")
print(f"Pico mensal de casos: {pico_casos['Data'].strftime('%m/%Y')} — {pico_casos['Casos']:,.0f} casos")
print(f"Pico mensal de óbitos: {pico_obitos['Data'].strftime('%m/%Y')} — {pico_obitos['Óbitos']:,.0f} óbitos")

fig, ax = plt.subplots(figsize=(14, 6))
ax.plot(p1_mensal["Data"], p1_mensal["Casos"], marker="o", label="Casos")
ax.plot(p1_mensal["Data"], p1_mensal["Óbitos"], marker="o", label="Óbitos")
ax.set_title("Evolução mensal de casos graves (SRAG) e óbitos por COVID-19 — 2020 a 2023")
ax.set_xlabel("Período")
ax.set_ylabel("Quantidade de registros")
ax.legend()
print("\n")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Interpretação dos resultados

Ao combinar dados anuais e mensais, a análise temporal mapeia a concentração de casos graves e óbitos por COVID-19 de 2020 a 2023 e destaca os picos de incidência do período, garantindo a interpretação dos volumes sem atribuição precipitada de causalidade.

## P2. Existem diferenças relevantes na letalidade por COVID-19 entre as regiões brasileiras?

A P2 avalia a letalidade agregada entre as regiões brasileiras calculando a razão entre óbitos e casos, o que permite comparar o impacto da doença independentemente das diferenças no tamanho das populações regionais.


```python
p2_regiao = (
    df_casos_obitos.groupby("Região", as_index=False)[["Casos", "Óbitos"]].sum()
)
p2_regiao["Letalidade (%)"] = np.where(
    p2_regiao["Casos"] > 0,
    p2_regiao["Óbitos"] / p2_regiao["Casos"] * 100,
    np.nan
)
p2_regiao["Participação dos casos (%)"] = p2_regiao["Casos"] / p2_regiao["Casos"].sum() * 100
p2_regiao["Participação dos óbitos (%)"] = p2_regiao["Óbitos"] / p2_regiao["Óbitos"].sum() * 100
p2_regiao = p2_regiao.sort_values("Letalidade (%)", ascending=False)

p2_regiao["Diferença para a menor letalidade (p.p.)"] = (
    p2_regiao["Letalidade (%)"] - p2_regiao["Letalidade (%)"].min()
)

display(p2_regiao.round(2))

plt.figure(figsize=(12, 6))
sns.barplot(data=p2_regiao, x="Região", y="Letalidade (%)")

print("\n")
plt.title("Letalidade agregada por COVID-19 segundo região — 2020 a 2023")
plt.xlabel("Região")
plt.ylabel("Letalidade (%)")
plt.tight_layout()
plt.show()
```

### Interpretação da P2

A análise regional contextualiza as variações de letalidade e o volume de ocorrências por meio de métricas em pontos percentuais, mantendo o caráter estritamente descritivo dos dados observados.

## P3. Como evoluiu o volume de doses de vacina aplicadas no Brasil ao longo do período analisado?

A P3 avalia a distribuição temporal das doses aplicadas de 2021 a 2023 por dados mensais e anuais, identificando os períodos de maior intensidade de vacinação sem inferir cobertura populacional ou número de pessoas imunizadas.

```python
p3_mensal = vacina_mensal_desc.copy()
p3_anual = (
    p3_mensal
    .groupby("Ano", as_index=False)["Total de Doses Aplicadas"]
    .sum()
)

# Participação de cada ano no total do período
p3_anual["Participação no total (%)"] = (
    p3_anual["Total de Doses Aplicadas"]
    / p3_anual["Total de Doses Aplicadas"].sum()
    * 100
)

# Variação percentual em relação ao ano anterior
p3_anual["Variação anual (%)"] = (
    p3_anual["Total de Doses Aplicadas"].pct_change() * 100
)

p3_anual["Variação anual (apresentação)"] = (
    p3_anual["Variação anual (%)"]
    .round(2)
    .astype("string")
)

# Como a vacinação teve início em 2021, não existe período
# anterior de vacinação dentro do recorte analisado.
primeiro_ano_vacinacao = p3_anual["Ano"].min()

p3_anual.loc[
    p3_anual["Ano"].eq(primeiro_ano_vacinacao),
    "Variação anual (apresentação)"
] = "Não se aplica"

print("Resumo anual da vacinação:")

display(
    p3_anual[
        [
            "Ano",
            "Total de Doses Aplicadas",
            "Participação no total (%)",
            "Variação anual (apresentação)"
        ]
    ].round(2)
)

p3_mensal["Data"] = pd.to_datetime(
    p3_mensal["Ano"].astype(str)
    + "-"
    + p3_mensal["Mês_num"].astype(str)
    + "-01",
    errors="coerce"
)

p3_mensal = p3_mensal.sort_values("Data")

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=p3_mensal,
    x="Data",
    y="Total de Doses Aplicadas",
    marker="o"
)

plt.title(
    "Evolução mensal das doses de vacina aplicadas — 2021 a 2023"
)

plt.xlabel("Período")
plt.ylabel("Total de doses aplicadas")

plt.xticks(rotation=45)

plt.ticklabel_format(
    style="plain",
    axis="y"
)
print("\n")
plt.tight_layout()
plt.show()
```

### Interpretação dos resultados

A análise temporal de 2021 a 2023 mapeia a distribuição mensal e anual das doses aplicadas, permitindo identificar os períodos de maior volume de vacinação sem determinar fatores causais ou inferir a cobertura vacinal da população.

### P5. Quais faixas etárias apresentaram maior letalidade entre os casos de SRAG por COVID-19?

A P5 avalia o impacto demográfico da doença combinando casos, óbitos e letalidade por faixa etária para mensurar com precisão as disparidades entre os grupos etários.

```python
# Define a ordem lógica das faixas etárias
ordem_faixas_letalidade = [
    "< 02 anos",
    "02 a 04 anos",
    "05 a 14 anos",
    "15 a 49 anos",
    "50 a 64 anos",
    "65 anos ou +"
]

# Agrega casos e óbitos por faixa etária
p5_faixa = (
    df_casos_obitos
    .groupby("Faixa Etária", as_index=False)[["Casos", "Óbitos"]]
    .sum()
)

# Calcula a letalidade somente quando existem casos registrados
p5_faixa["Letalidade (%)"] = np.where(
    p5_faixa["Casos"] > 0,
    p5_faixa["Óbitos"] / p5_faixa["Casos"] * 100,
    np.nan
)

# Calcula a participação de cada faixa etária no total de casos
p5_faixa["Participação dos casos (%)"] = (
    p5_faixa["Casos"]
    / p5_faixa["Casos"].sum()
    * 100
)

# Calcula a participação de cada faixa etária no total de óbitos
p5_faixa["Participação dos óbitos (%)"] = (
    p5_faixa["Óbitos"]
    / p5_faixa["Óbitos"].sum()
    * 100
)

# Define a ordem das faixas etárias
p5_faixa["Faixa Etária"] = pd.Categorical(
    p5_faixa["Faixa Etária"],
    categories=ordem_faixas_letalidade,
    ordered=True
)

# Ordena a tabela
p5_faixa = p5_faixa.sort_values("Faixa Etária")

# Exibe a tabela consolidada
print("Letalidade e distribuição de casos e óbitos por faixa etária:")

display(
    p5_faixa[
        [
            "Faixa Etária",
            "Casos",
            "Óbitos",
            "Letalidade (%)",
            "Participação dos casos (%)",
            "Participação dos óbitos (%)"
        ]
    ].round(2)
)

# Gráfico
plt.figure(figsize=(12, 6))

sns.barplot(
    data=p5_faixa,
    x="Faixa Etária",
    y="Letalidade (%)"
)

plt.title(
    "Letalidade observada por COVID-19 segundo a faixa etária — 2020 a 2023"
)

plt.xlabel("Faixa etária")
plt.ylabel("Letalidade (%)")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
print("\n")
plt.show()
```

### Interpretação da P5

A interpretação contrapõe a letalidade ao volume absoluto de mortes, ressaltando o impacto do tamanho da amostra sobre a estabilidade dos percentuais e limitando a análise aos valores calculados sem inferir significância estatística.

### P6. Há indícios, ao observar o volume de doses aplicadas e a letalidade por faixa etária, de algum padrão de associação entre essas variáveis?

A P6 investiga a relação entre o volume de doses aplicadas e a letalidade por faixa etária por meio do reagrupamento e cruzamento de bases, utilizando gráficos de dispersão e correlações exploratórias para identificar padrões agregados com cautela metodológica.

```python
# Copia e prepara os dados de vacinação
p6_vac = df_vacina_idade.reset_index().copy()

p6_vac = p6_vac.rename(
    columns={"Total de Doses Aplicadas": "Doses"}
)

# Junta vacinação com casos, óbitos e letalidade
p6 = pd.merge(
    p6_vac,
    p5_faixa.astype({"Faixa Etária": "string"})[
        ["Faixa Etária", "Casos", "Óbitos", "Letalidade (%)"]
    ],
    on="Faixa Etária",
    how="left",
    validate="one_to_one"
)

# Remove faixas sem casos registrados
# Ausência de denominador não é convertida em letalidade = 0
p6 = p6[
    p6["Casos"].notna() & (p6["Casos"] > 0)
].copy()

# Define a ordem correta das faixas etárias
p6["Faixa Etária"] = pd.Categorical(
    p6["Faixa Etária"],
    categories=ordem_faixas_letalidade,
    ordered=True
)

# Ordena
p6 = p6.sort_values("Faixa Etária")

# Participação das doses dentro da comparação
p6["Participação das doses (%)"] = (
    p6["Doses"] / p6["Doses"].sum() * 100
)

# Exibe a tabela
display(p6.round(2))


# ============================================================
# CORRELAÇÕES
# ============================================================

# Associação exploratória entre doses e letalidade.
# O número de observações é pequeno e as variáveis são agregadas
# por faixa etária; portanto, a correlação não deve ser interpretada
# como causalidade.

if len(p6) >= 3:

    correlacao_pearson = p6["Doses"].corr(
        p6["Letalidade (%)"],
        method="pearson"
    )

    correlacao_spearman = p6["Doses"].corr(
        p6["Letalidade (%)"],
        method="spearman"
    )

    print("\n")
    print(
        f"Correlação de Pearson (exploratória): "
        f"{correlacao_pearson:.3f}"
    )

    print(
        f"Correlação de Spearman (exploratória): "
        f"{correlacao_spearman:.3f}"
    )

else:
    print("Não há observações suficientes para calcular correlação.")


# ============================================================
# GRÁFICO: DOSES DE VACINA × LETALIDADE
# ============================================================

fig, ax1 = plt.subplots(figsize=(14, 7))

# Barras: doses aplicadas
sns.barplot(
    x="Faixa Etária",
    y="Doses",
    data=p6,
    ax=ax1,
    hue="Faixa Etária",
    legend=False,
    palette="mako",
    alpha=0.7
)

ax1.set_xlabel("Faixa Etária")
ax1.set_ylabel(
    "Total de Doses Aplicadas",
    color="tab:blue"
)

ax1.tick_params(
    axis="y",
    labelcolor="tab:blue"
)

ax1.ticklabel_format(
    style="plain",
    axis="y"
)

ax1.set_title(
    "Doses de Vacina Aplicadas e Letalidade por Faixa Etária"
)

plt.xticks(
    rotation=45,
    ha="right"
)


# Segundo eixo: letalidade
ax2 = ax1.twinx()

sns.lineplot(
    x="Faixa Etária",
    y="Letalidade (%)",
    data=p6,
    ax=ax2,
    color="red",
    marker="o",
    sort=False
)

ax2.set_ylabel(
    "Letalidade (%)",
    color="tab:red"
)

ax2.tick_params(
    axis="y",
    labelcolor="tab:red"
)

plt.tight_layout()

print("\n")
plt.show()
```

### Interpretação da P6

Ao integrar a análise visual e as métricas de associação, a interpretação da P6 expõe a relação entre o volume de doses e a letalidade nas diferentes faixas etárias, sugerindo um padrão exploratório de mitigação de risco que, por ausência de acompanhamento individual e limite amostral, não comprova relação de causa e efeito.

## 5. Limitações do estudo

As principais limitações identificadas ao longo dos três artefatos são:
   

#### 5.1 - Casos analisados

A base de casos corresponde a registros de casos graves de SRAG por COVID-19, não à totalidade de infecções por COVID-19 na população. Portanto, os indicadores não devem ser interpretados como incidência ou letalidade de todos os casos da doença.
   
    

#### 5.2 - Dados agregados

As bases não possuem vínculo individual entre vacinação e ocorrência de SRAG ou óbito. Assim, a P6 compara grupos agregados e não indivíduos.
   
     
   
#### 5.3 - Comparabilidade das faixas etárias

A base de vacinação possui uma classificação etária diferente da base de casos e óbitos. Foi necessário realizar um reagrupamento para viabilizar a comparação. Essa transformação reduz a precisão da correspondência entre as categorias originais e deve ser considerada na interpretação da P6.
   
     

#### 5.4 - Doses aplicadas não equivalem a cobertura vacinal

O número de doses aplicadas não representa diretamente o percentual da população vacinada. Para estimar cobertura seria necessária uma população de referência e uma definição clara do esquema vacinal considerado.
   
      

#### 5.5 - Interpretação causal

As análises são predominantemente descritivas e exploratórias. Associação estatística entre variáveis agregadas não é suficiente para demonstrar causalidade.
   
      

#### 5.6 - Período de análise

O recorte foi definido em função do escopo do projeto: 2020–2023 para casos/óbitos e 2021–2023 para vacinação. A ausência de períodos anteriores limita algumas comparações de variação.
